## 2. The question: decision, action, cost of a wrong call

Decision it improves: Which pages a content or SEO team spends limited review time on this cycle, out of a pool this large (30,000 items, 54.2% currently declining). Unit of analysis: one content item (content_id). Output: a ranked refresh queue with scores and reason codes. Action someone takes: Pull the top-ranked items into that queue — update, expand, or re-optimize them first. ## 3. Quick look at the data (2-3 real numbers)

All 30,000 rows in the starter dataset pass the filters (impressions_90d > 0, content_age_days >= 90), so this lane has a full, usable pool to work with. Just over half (54.2%) of pages are trending down, and the typical page has meaningful existing visibility (median 731 impressions over 90 days) and is over 7 months old (median 236 days) — old enough to plausibly need a refresh, not brand-new content still ramping up.

```python
import pandas as pd

url = "https://raw.githubusercontent.com/GauriNehe/Flyrank-ml-internship/main/data/raw/content_refresh_anonymized.csv"
df = pd.read_csv(url)


df_f = df[(df["impressions_90d"] > 0) & (df["content_age_days"] >= 90)].drop_duplicates(subset="content_id")

print("Total rows (raw):", len(df))
print("Rows after filtering:", len(df_f))
print("Share trending down:", (df_f["trend_direction"] == "down").mean().round(3))
print("Median impressions_90d:", df_f["impressions_90d"].median())
print("Median content_age_days:", df_f["content_age_days"].median())

```
